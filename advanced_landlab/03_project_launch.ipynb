## Scope examples

- How does hillslope diffusivity affect relief or drainage density?
- How does erodibility affect channel profiles at a fixed model time?
- How does hydraulic conductivity alter runoff or stored surface water?
- How does opening only one boundary change drainage organization?
- How does a synthetic surface compare with one supplied DEM under the same
  component setup?

Phrase the question so it can be answered by comparing two to four runs.

## Build the smallest baseline

Copy only the cells required for one known-working run from the fundamentals or
advanced notebook. Restart the kernel and run all cells. Save the baseline
diagnostic before changing a parameter.

Use this debugging order:

1. Were setup cells run in order after the last restart?
2. Are required field names present at the correct grid element?
3. Are units internally consistent?
4. Are boundary conditions appropriate?
5. Is the timestep appropriate, and is component order defensible?
6. Does the diagnostic answer the question?
