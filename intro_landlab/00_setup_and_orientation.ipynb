## 2. Create and plot a small grid

```python
from landlab import RasterModelGrid, imshow_grid

grid = RasterModelGrid((3, 4), xy_spacing=10.0)
z = grid.add_zeros("topographic__elevation", at="node")
z[:] = grid.x_of_node + 0.5 * grid.y_of_node

print(grid)
imshow_grid(grid, "topographic__elevation", colorbar_label="Elevation (m)")
```
